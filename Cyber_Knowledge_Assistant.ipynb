```python
files = [
    f for f in sorted(glob.glob("knowledge/**/*.md", recursive=True))
    if os.path.basename(f) not in ("index.md", "log.md")
]

problems = 0

for path in files:
    with open(path, encoding="utf-8") as f:
        text = f.read()

    missing = []

    if not re.search(r"(?m)^type\s*:", text):
        missing.append("type: line")

    if not re.search(r"(?m)^#\s+.+", text):
        missing.append("a # title")

    if not re.search(r"(?m)^##\s+.+", text):
        missing.append("at least one ## section")

    if missing:
        problems += 1
        print("PROBLEM", path, "is missing", ", ".join(missing))
    else:
        print("OK", path)

index_text = (
    "---\ntype: Index\ntitle: Cybersecurity Knowledge Index\n---\n\n"
    "# Cybersecurity Knowledge Index\n"
    + "\n".join(f"- {path.replace('knowledge/', '')}" for path in files)
    + "\n"
)

with open("knowledge/index.md", "w", encoding="utf-8") as f:
    f.write(index_text)

print("\nAll notes are good" if problems == 0
      else f"\n{problems} note(s) need fixing")
print("Saved knowledge/index.md")
```

```python
import os
import re
import glob
import numpy as np
from sentence_transformers import SentenceTransformer

# Find all cybersecurity Markdown notes
files = [
    f for f in sorted(glob.glob("knowledge/**/*.md", recursive=True))
    if os.path.basename(f) not in ("index.md", "log.md")
]

# Load the embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Split notes into sections and store their text
chunks = []

for path in files:
    with open(path, "r", encoding="utf-8") as f:
        content = f.read()

    # Split the note whenever a ## heading appears
    sections = re.split(r"(?m)(?=^##\s+)", content)

    for section in sections:
        section = section.strip()

        if section:
            chunks.append({
                "source": path,
                "text": section
            })

# Create embeddings for all sections
texts = [chunk["text"] for chunk in chunks]

if not texts:
    raise ValueError(
        "No note sections found. Run Cell 2 and Cell 3 first."
    )

embeddings = model.encode(
    texts,
    convert_to_numpy=True,
    normalize_embeddings=True
)

print("Cybersecurity notes loaded:", len(files))
print("Text sections created:", len(chunks))
print("Embedding matrix shape:", embeddings.shape)
print("Cell 4 completed successfully!")
```

```python
def search_knowledge(question, top_k=3):
    # Convert the user's question into an embedding
    query_embedding = model.encode(
        [question],
        convert_to_numpy=True,
        normalize_embeddings=True
    )[0]

    # Calculate similarity scores against all note sections
    scores = embeddings @ query_embedding

    # Get the highest-scoring sections
    top_indices = np.argsort(scores)[::-1][:top_k]

    results = []

    for index in top_indices:
        results.append({
            "source": chunks[index]["source"],
            "text": chunks[index]["text"],
            "score": float(scores[index])
        })

    return results


# Test the search function
question = "How can I protect my computer from malware?"
results = search_knowledge(question)

print("Question:", question)
print("=" * 60)

for result in results:
    print("Source:", result["source"])
    print("Similarity score:", round(result["score"], 3))
    print(result["text"][:700])
    print("-" * 60)
```

```python
import re
import os

def ask_cyber_assistant(question, top_k=8):
    question = question.strip()

    if not question:
        return "Please enter a cybersecurity question."

    # Search the existing knowledge base
    results = search_knowledge(question, top_k=top_k)

    if not results:
        return "No relevant information was found in the knowledge files."

    # Keep only reasonably relevant results
    relevant_results = [
        result for result in results
        if result["score"] >= 0.20
    ]

    if not relevant_results:
        return (
            "I couldn't find enough relevant information in your "
            "12 cybersecurity knowledge files. Try a more specific question."
        )

    sections = []
    seen = set()

    for result in relevant_results:
        text = result["text"].strip()

        # Remove YAML metadata at the start of a note
        text = re.sub(
            r"\A---\s*\n.*?\n---\s*\n?",
            "",
            text,
            flags=re.DOTALL
        )

        # Remove type, title, and source metadata
        text = re.sub(
            r"(?im)^\s*(type|title|source)\s*:\s*.*$",
            "",
            text
        )

        # Clean up Markdown headings and separators
        text = re.sub(r"(?m)^#{1,6}\s*", "", text)
        text = re.sub(r"(?m)^---\s*$", "", text)
        text = text.strip()

        if text and text not in seen:
            sections.append(text)
            seen.add(text)

    if not sections:
        return "No useful text was found in the retrieved notes."

    # Return all relevant information without exposing source metadata
    answer = "Here is the information from your cybersecurity notes:\n\n"

    for section in sections:
        answer += section + "\n\n"

    answer += (
        "These details are based on the information available "
        "in your knowledge files. Ask a follow-up question for more detail."
    )

    return answer


# Test with a short question
print(ask_cyber_assistant("What is ransomware?"))
```

```python
import os
import zipfile
from google.colab import files

# Name of the ZIP file to create
zip_filename = "Cyber_Knowledge_Assistant.zip"

# Create a ZIP archive containing your knowledge folder
with zipfile.ZipFile(zip_filename, "w", zipfile.ZIP_DEFLATED) as zipf:
    for root, dirs, filenames in os.walk("knowledge"):
        for filename in filenames:
            file_path = os.path.join(root, filename)

            # Keep the knowledge/ folder structure inside the ZIP
            zipf.write(file_path, arcname=file_path)

print("ZIP file created:", zip_filename)
print("Files included:")

for root, dirs, filenames in os.walk("knowledge"):
    for filename in filenames:
        print(os.path.join(root, filename))
```

```python
import gradio as gr

# Connect the chatbot interface to your assistant
def chatbot_response(message, history):
    if not message or not message.strip():
        return "Please enter a cybersecurity question."

    return ask_cyber_assistant(message.strip())


# Create the chatbot webpage
demo = gr.ChatInterface(
    fn=chatbot_response,
    title="Cybersecurity Knowledge Assistant",
    description=(
        "Ask questions about malware, passwords, network security, "
        "ransomware, online fraud, web security, mobile security, "
        "data privacy, ethical hacking, digital forensics, "
        "network attacks, and incident response."
    ),
    examples=[
        "How can I protect my computer from malware?",
        "How can I create a strong password?",
        "What should I do if I suspect online fraud?",
        "How can I protect my mobile phone?",
        "What is a DDoS attack?",
        "What should I do after a cybersecurity incident?"
    ]
)

# Launch the chatbot
demo.launch(share=True)
```
